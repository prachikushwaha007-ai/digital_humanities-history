```python
# TO OPEN BOOK URL THROUGH PANDAS
book_url='https://gutenberg.org/cache/epub/38253/pg38253.txt'
```

```python
import urllib.request
```

```python
import pandas as pd
```

```python
with urllib.request.urlopen(book_url) as response:
  raw_text=response.read().decode('utf-8')
```

```python
lines = raw_text.split('\n')
```

```python
df_history = pd.DataFrame(lines, columns=['book_text'])
```

```python
df_history['book_text'] = df_history['book_text'].replace('\r','')
```
