```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
df = pd.read_csv(url)
```

```python
df
```

```python
# Filter the dataFrame.
df.filter(['Name','Age'])
```

```python
df.filter(['Age', 'Name'])#
```

```python
# creating a new .csv by filter.
new_df = df.filter(['Age', 'Name'])
```

```python
new_df
```

```python
# boolean for True if column name contains individual.
df.filter(['Name']).Name.str.contains('Miss.')
```

```python
# value_counts for very quick quantitative and qualitative analysis.
df.filter(["Name"]).Name.str.contains("Miss.").value_counts()
```

```python
df[df.filter(["Age","Name"]).Name.str.contains("Miss.") == True]
```

```python
# only show a name and age if it contains Miss.
df[df.Name.str.contains("Miss.") ==True].filter(["Name","Age"])
```

```python
#Query
df.query("Pclass == 3")
```

```python
df.query("Pclass == 3 & Survived == 1")
```

```python
df.query("Pclass==3 & Survived==1 & Age > 40")
```

```python
# if we use or | then use it with parenthesis.
df.query("Pclass==3 & Survived==1 & (Age > 40 | Age <10)")
```
