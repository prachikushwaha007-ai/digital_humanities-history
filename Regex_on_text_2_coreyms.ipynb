```python
import re
```

```python
# 1 \.   - Any Character Except New line
# 2 \d   - Digit(0-9) , DATE FIND- 18\d\d pattern
# 3 \D   - Not a Digit (0-9)
# 4 \w   - Word Character(a-z, A-Z, 0-9, _), STYLOMETRY-NLP
# 5 \W   - Not a Word Character
# 6 \s   - Whitespace (space, tab, newline)
# 7 \S   - Not Whitespace (space, tab, newline)
```

```python
# 08 \b   - Word Boundary , \bruler\b pattern find
# 09 \B   - Not a Word Boundary
# 10 ˆ    - Beginning of a String
# 11 $    - End of a String
# 12 []   - OCR Mistake in spell (Centr[ee]r)
# 13 [ˆ ] - Matches Characters NOT in Brackets
# 14 [a-zA-1-9]  -  RANGE SET
```

```python
# Quantifiers.
# *     - 0 or more
# +     - 1 or more
# ?     - 0 or one
# {3}   - Exact Number
# {3,4} - Range of Numbers (Minimum, Maximum)
```

```python
emails = '''
CoreyMSchafer@gmail.com
corey.schafer@university.edu
corey-321-schafer@my-work.net
'''
```

```python
# to find emails- first match the corey in upper or lower case by [a-zA-Z]
# to add letters after @ using+ and adding all cases a-zA-Z and digits 0-9 and . and -
#to add . Use \. And using upward slash | addresses.

pattern = re.compile(r'[a-zA-Z0-9_+.-]+@[a-zA-Z-]+\.[a-zA-Z0-9-.]+')

matches = pattern.finditer(emails)

for match in matches:
    print(match)
```

```python
# Matching emails via word character.
# for print, use group() method.

pattern = re.compile(r'[\w.+-]+@[\w-]+[\w.-]+')

matches = pattern.finditer(emails)

for match in matches:
    print(match.group(0))
```

```python
urls = '''
https://www.google.com
http://coreyms.com
https://YouTube.com
https://www.isro.gov
'''
```

```python
# Matching urls by group method ()

pattern = re.compile(r'https?://(www\.)?(\w+)(\.\w+)')

matches = pattern.finditer(urls)

for match in matches:
  print(match.group(0))
```

```python
# Print urls in high domain.

subbed_urls = pattern.sub(r'\2\3', urls)
print(subbed_urls)
```
