```python
text_to_search = '''
abcdefghijklmnopqrstuvwxyz
ABCDEFGHIJKLMNOPQRSTUVWXYZ
1234567890

Ha HaHa

MetaCharacters(Need to be escaped):
. ˆ $ * + ? { } [ ] \ | ( )

coreyms.com

321-555-4321
123.555.1234

Mr. Schafer
Mr Smith
Ms Davis
Mrs. Robinson
Mr. T
'''
```

```python
sentence = 'Start a sentence and then bring it to an end'
```

```python
print('\tTab')
```

```python
print(r'\tTab') # r = raw text
```

```python
# re.compile() method itext_to_searchs used to seperate out patterns into a variable & also reuse that variable for multiple searches
pattern = re.compile(r'abc')
matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
print(text_to_search[1:4]) # span is 1 to 4 in above output.
```

```python
pattern = re.compile(r'\D')  # \ for escaping MetaCharacters.
matches = pattern.finditer(text_to_search)

for match in matches:
  print(match)
```

```python
pattern = re.compile(r'coreyms\.com')
matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
# 1 \.   - Any Character Except New line
# 2 \d   - Digit(0-9) , DATE FIND- 18\d\d pattern
# 3 \D   - Not a Digit (0-9)
# 4 \w   - Word Character(a-z, A-Z, 0-9, _), STYLOMETRY-NLP
# 5 \W   - Not a Word Character
# 6 \s   - Whitespace (space, tab, newline)
# 7 \S   - Not Whitespace (space, tab, newline)
```

```python
# 08 \b   - Word Boundary , \bruler\b pattern find
# 09 \B   - Not a Word Boundary
# 10 ˆ    - Beginning of a String
# 11 $    - End of a String
# 12 []   - OCR Mistake in spell (Centr[ee]r)
# 13 [ˆ ] - Matches Characters NOT in Brackets
# 14 [a-zA-1-9]  -  RANGE SET
```

```python
pattern = re.compile(r'\.')
matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
pattern = re.compile(r'\d\w')
matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
pattern = re.compile(r'end$')

matches = pattern.finditer(sentence)

for match in matches:
    print(match)
```

```python
# Matching digits.
pattern = re.compile(r'\d\d\d.\d\d\d.\d\d\d\d')

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
# Opening the text file 'freedom of the seas'.
with open('/content/drive/MyDrive/freedom of the seas .txt', 'r', encoding='utf-8') as f:
    contents = f.read()

matches = pattern.finditer(contents)

for match in matches:
   print(match)
```

```python
# find the specific words by using word boundary and findall and matches the length.
# dutch
pattern = re.compile(r'\bdutch\b', re.IGNORECASE)
matches = pattern.findall(contents)
print(matches)
len(matches)
```

```python
# Portuguese
pattern = re.compile(r'\bportuguese\b', re.IGNORECASE)
matches = pattern.findall(contents)
print(matches)
len(matches)
```

```python
# kling
pattern = re.compile(r'\bkling\b', re.IGNORECASE)
matches = pattern.findall(contents)
print(matches)
len(matches)
```

```python
# tax
pattern = re.compile(r'\btax\b', re.IGNORECASE)
matches = pattern.findall(contents)
print(matches)
len(matches)
```
