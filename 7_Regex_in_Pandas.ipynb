```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
```

```python
df = pd.read_csv(url)
```

```python
df.head(891)
```

```python
names = df.Name.tolist()
names
```

```python
revs = []
for name in names:
  if 'Rev.' in name:
    revs.append(name)
print(revs)
```

```python
# for seeing the OTHER inf. about Revs.as df.
df.loc[df['Name'].str.contains('Rev.')]
```

```python
# To find DF except 'Rev.'.
df.loc[~df['Name'].str.contains('Rev.')]
```

```python
# REGEX Based.
df.loc[(df['Name'].str.contains('Rev\.'))|(df['Name'].str.contains('Mr\.'))]
```

```python
# REGEX IN SHORT WAY.
df.loc[df['Name'].str.contains('Rev\.|Mr\.', regex=True)]
```

```python
# regex & case.
df.loc[df['Name'].str.contains('Al',case=False|True, regex=True)]
```

```python
df['Name'] = df['Name'].str.replace('Rev.','Cev',regex=False)
df['Name']
```
